```python
#import lib that need in this project:
# pandas for data manipulation and numpy is for mathematical function
import pandas as pd
import numpy as np
```

```python
#file save in _file variable  
INPUT_FILE = "DATA_SEICNCE/train.csv"
OUTPUT_FILE = "cleaned_data.csv"
LOG_FILE = "cleaning_log.txt"
```

```python
# and then we load in dataframe to us of pandas for data more manipulation
df = pd.read_csv(INPUT_FILE)
log = []

log.append(f"Initial shape: {df.shape}")
```

```python
# Standardize column names then replace or remove unwented thing which easy to see in table 
df.columns = (
    df.columns.str.lower()
        .str.strip()
        .str.replace(" ", "_")
        .str.replace(r"[^a-z0-9_]", "", regex=True)
)

log.append("Column names standardized")
```

```python
# Remove duplicates 
dup_count = df.duplicated().sum()
df.drop_duplicates(inplace=True)
log.append(f"Removed {dup_count} duplicate rows")
```

```python
#  Fix numeric data types 
for col in df.columns:
    if df[col].dtype == "object":
        converted = pd.to_numeric(df[col], errors="coerce")
        if converted.notna().sum() > 0.7 * df[col].notna().sum():
            df[col] = converted
            log.append(f"Converted {col} to numeric")
```

```python
#  Parse date columns (safe) 
for col in df.columns:
    if df[col].dtype == "object":
        parsed = pd.to_datetime(df[col], errors="coerce")
        if parsed.notna().sum() > 0.7 * df[col].notna().sum():
            df[col] = parsed
            log.append(f"Parsed {col} as date")
```

```python
#  Handle missing values
for col in df.columns:
    missing = df[col].isnull().sum()
    if missing > 0:
        if df[col].dtype in ["float64", "int64"]:
            df[col].fillna(df[col].mean(), inplace=True)
            log.append(f"Filled {missing} missing values in {col} with mean")
        else:
            df[col].fillna(df[col].mode()[0], inplace=True)
            log.append(f"Filled {missing} missing values in {col} with mode")
```

```python
# ---------- Save cleaned data ----------
df.to_csv(OUTPUT_FILE, index=False)
```

```python

#  Save cleaning log
with open(LOG_FILE, "w") as f:
    for line in log:
        f.write(line + "\n")

print(" Data cleaning completed successfully!")
print(" Cleaned file:", OUTPUT_FILE)
print(" Log file:", LOG_FILE)
```
